```python
# internal modules
from color_brains.data.sequential_data import SequentialData
import pandas as pd
import numpy as np
from typing import *
```

```python
sd = SequentialData()
df = sd.dataset
feature_df = df
df.head()
```

```python
rgb_columns = [c for c in df.columns if c != sd.CmapNameColumn]
def mangle_rgb_names(tag: str) -> List[str]:
    return [f'{tag}_{column}' for column in rgb_columns]
rgb_columns
```

```python
# starting rgb
prev_rgb = feature_df.groupby([sd.CmapNameColumn]).shift(1).reset_index(drop=True)
prev_rgb.columns = mangle_rgb_names("previous")
feature_df = feature_df.join(prev_rgb)
feature_df.head()
```

```python
# squareroots of current rgb 
sqrt_df =  np.sqrt(feature_df[rgb_columns])
sqrt_df.columns = mangle_rgb_names("sqrt")
feature_df = feature_df.join(sqrt_df)

# cuberoots of current rgb
cuberoot_df =  np.sqrt(feature_df[rgb_columns])
cuberoot_df.columns = mangle_rgb_names("cbrt")
feature_df = feature_df.join(cuberoot_df)

feature_df.head()
```

```python
# rgb of start of sequence
start_rgb = df.groupby(sd.CmapNameColumn).nth(0)
start_rgb = start_rgb.set_index(sd.CmapNameColumn)
start_rgb.columns = mangle_rgb_names("first")
feature_df = pd.merge(feature_df, start_rgb, how='left', on=[sd.CmapNameColumn])
feature_df.head()
```
